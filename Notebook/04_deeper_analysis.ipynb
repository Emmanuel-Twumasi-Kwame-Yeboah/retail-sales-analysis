## Practical 1

Question 1: What is the total revenue?

```python
total_revenue = df["Revenue"].sum()
total_revenue
```

Interpretation: The total revenue generated from all 500 transactions was 64,463.98. This represents the combined revenue from all products and countries in the dataset.

Question 2: What is the average transaction value?

```python
average_transaction_value = df["Revenue"].mean()
average_transaction_value
```

Interpretation: The average transaction value was 128.93. This means that, on average, each transaction generated approximately 128.93 in revenue.

Question 3: Which 10 products generate the most revenue?

```python
top_10_products = (
    df.groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_10_products
```

Interpretation: POSTAGE generated the highest revenue at 18,414.00, followed by REGENCY CAKESTAND 3 TIER at 11,959.50 and SET 7 BABUSHKA NESTING BOXES at 8,126.00. The results show that revenue was concentrated among a small number of products, with POSTAGE contributing the highest amount.

Question 4: Which 5 countries generate the most revenue?

```python
top_5_countries = (
    df.groupby("Country")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

top_5_countries
```

Interpretation: The United Kingdom generated the highest revenue at 52,766.81, far exceeding the other four countries. Germany ranked second with 2,784.23, while the Netherlands recorded the lowest revenue among the top five at 1,380.17.

Question 5: Which products sell the highest quantities?

```python
top_10_quantity_products = (
    df.groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_10_quantity_products
```

Interpretation: PACK OF 72 RETROSPOT CAKE CASES had the highest quantity sold, with 4,353 units, followed by POPCORN HOLDER with 3,566 units. These products had the highest sales volume in terms of quantity within the dataset.

Question 6: Which month generated the highest revenue?

```python
monthly_revenue = (
    df.groupby("Month")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

monthly_revenue
```

Interpretation: June generated the highest revenue at 7,237.98, while February generated the lowest revenue at 3,409.79. Revenue varied across the months, with June showing the strongest monthly performance.
